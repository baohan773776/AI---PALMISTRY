```python
from pathlib import Path
from ultralytics import YOLO

PROJECT_DIR = Path(r"D:\UEH\2 - HKĐ2026\AI\BUỔI 4\PALM")

candidates = [
    PROJECT_DIR / "runs" / "detect" / "palm_finetune_fate" / "weights" / "best.pt",
    PROJECT_DIR / "runs" / "detect" / "palm_finetune_fate" / "weights" / "last.pt",
    PROJECT_DIR / "runs" / "detect" / "palmistry_yolov8_cpu" / "weights" / "best.pt",
    PROJECT_DIR / "runs" / "detect" / "palmistry_yolov8_cpu" / "weights" / "last.pt",
    PROJECT_DIR / "runs" / "detect" / "palmistry_yolov8_cpu_resume" / "weights" / "best.pt",
    PROJECT_DIR / "runs" / "detect" / "palmistry_yolov8_cpu_resume" / "weights" / "last.pt",
]

weights_path = next((p for p in candidates if p.exists()), None)
if weights_path is None:
    searched = "\n".join(str(p) for p in candidates)
    raise FileNotFoundError(f"Không tìm thấy weight palmistry. Đã tìm:\n{searched}")

model = YOLO(str(weights_path))
expected_classes = {"fate", "head", "heart", "life"}
model_classes = set(model.names.values())

if model_classes != expected_classes:
    raise ValueError(
        "Đang load nhầm model không phải palmistry. "
        f"Classes hiện tại: {model.names}. Weight: {weights_path}"
    )

print("Loaded palmistry model:", weights_path)
print("Classes:", model.names)
```
